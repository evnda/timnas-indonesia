```python
import os
import glob
import subprocess
import pandas as pd

OUTPUT_DIR = "/content/tweets-data"
os.makedirs(OUTPUT_DIR, exist_ok=True)

print("Folder penyimpanan siap:", OUTPUT_DIR)
```

```python
from getpass import getpass

auth_token = getpass("Masukkan auth_token X kamu: ")

keywords = [
    "Timnas",
    '"Timnas Indonesia"',
    '"Timnas Garuda"',
    "TimnasDay",
    '"sepak bola Indonesia"'
]

for i, keyword in enumerate(keywords, start=1):
    print(f"\n{'='*50}")
    print(f"Crawling {i}/{len(keywords)}: {keyword}")
    print("="*50)

    command = [
        "npx", "--yes", "tweet-harvest@2.7.1",
        "--token", auth_token,
        "--search-keyword", keyword,
        "--limit", "60",
        "--delay", "3",
        "--output-filename", f"timnas_{i}",
        "--export-format", "csv",
        "--tab", "LATEST"
    ]

    result = subprocess.run(
        command,
        cwd=OUTPUT_DIR,
        capture_output=True,
        text=True
    )

    print(result.stdout)

    if result.stderr.strip():
        print("Pesan tambahan:", result.stderr[-1500:])

auth_token = None
print("\nProses crawling untuk semua kata kunci selesai.")
```

```python
import os
import glob
import pandas as pd

OUTPUT_DIR = "/content/tweets-data"
RAW_PATH = os.path.join(OUTPUT_DIR, "dataset_timnas_raw.csv")

# Ambil semua CSV, kecuali dataset gabungan yang mungkin sudah ada
files = glob.glob(
    os.path.join(OUTPUT_DIR, "**", "*.csv"),
    recursive=True
)

files = [
    f for f in files
    if os.path.abspath(f) != os.path.abspath(RAW_PATH)
    and os.path.basename(f) != "dataset_timnas.csv"
]

dataframes = []

for file in files:
    try:
        df = pd.read_csv(file)

        if not df.empty:
            df["sumber_file"] = os.path.basename(file)
            dataframes.append(df)

    except pd.errors.EmptyDataError:
        pass

if dataframes:
    raw = pd.concat(dataframes, ignore_index=True)

    # Hapus duplikat berdasarkan ID jika tersedia
    id_col = next(
        (c for c in ["id_str", "tweet_id", "id"]
         if c in raw.columns),
        None
    )

    if id_col:
        raw = raw.drop_duplicates(subset=id_col)
    else:
        raw = raw.drop_duplicates()

    raw.to_csv(RAW_PATH, index=False, encoding="utf-8-sig")

    print("Dataset mentah berhasil disimpan.")
    print("Jumlah posting unik:", len(raw))
    print("Lokasi:", RAW_PATH)
else:
    raise ValueError(
        "Tidak ada CSV berisi data. Proses crawling belum menghasilkan posting."
    )
```

```python
import re
import pandas as pd
import os

# Membaca hasil crawling
df = pd.read_csv("/content/tweets-data/dataset_timnas_raw.csv")

# Cari kolom yang sesuai dengan data hasil crawling
def find_column(candidates, columns):
    lookup = {c.lower(): c for c in columns}
    for candidate in candidates:
        if candidate.lower() in lookup:
            return lookup[candidate.lower()]
    return None

text_col = find_column(
    ["full_text", "text", "tweet", "content"],
    df.columns
)
date_col = find_column(
    ["created_at", "date", "created", "timestamp"],
    df.columns
)
like_col = find_column(
    ["like_count", "favorite_count", "likes", "likeCount"],
    df.columns
)

if text_col is None:
    raise ValueError(
        "Kolom teks tidak ditemukan. Periksa kolom dataset mentah."
    )

hasil = pd.DataFrame()
hasil["text_raw"] = df[text_col].fillna("").astype(str)

hasil["created_at"] = (
    df[date_col] if date_col else pd.NA
)

hasil["like_count"] = (
    pd.to_numeric(df[like_col], errors="coerce")
    if like_col else pd.NA
)

# Buang teks kosong
hasil = hasil[hasil["text_raw"].str.strip() != ""].copy()

# Bersihkan teks tanpa menghapus kata negasi
def clean_text(text):
    text = text.lower()
    text = re.sub(r"https?://\S+|www\.\S+", " ", text)
    text = re.sub(r"@\w+", " ", text)
    text = re.sub(r"#", "", text)
    text = re.sub(r"[^\w\s]", " ", text)
    text = re.sub(r"\s+", " ", text).strip()
    return text

hasil["text"] = hasil["text_raw"].apply(clean_text)

# Buang teks yang kosong setelah dibersihkan
hasil = hasil[hasil["text"] != ""].copy()

# Hapus duplikat teks setelah pembersihan
hasil = hasil.drop_duplicates(subset=["text"])

# Pastikan jumlah suka berupa angka
hasil["like_count"] = pd.to_numeric(
    hasil["like_count"], errors="coerce"
)

# Susun kolom untuk tugas
hasil = hasil[
    ["text_raw", "text", "created_at", "like_count"]
]

CLEAN_PATH = "/content/tweets-data/dataset_timnas.csv"
hasil.to_csv(CLEAN_PATH, index=False, encoding="utf-8-sig")

print("Preprocessing selesai.")
print("Jumlah data bersih:", len(hasil))
print("Dataset bersih:", CLEAN_PATH)
```

```python

import pandas as pd

file = "/content/tweets-data/dataset_timnas.csv"
df = pd.read_csv(file)

print("===== 1. PEMERIKSAAN DUPLIKAT =====")
print("Duplikat berdasarkan teks bersih:", df["text"].duplicated().sum())

print("\n===== 2. PEMERIKSAAN DATA KOSONG =====")
print(df[["text_raw", "text", "created_at", "like_count"]].isna().sum())

print("\n===== 3. PEMERIKSAAN JUMLAH SUKA =====")
print("Jumlah suka bukan angka:",
      pd.to_numeric(df["like_count"], errors="coerce").isna().sum())

print("\n===== 4. PEMERIKSAAN KATA NEGASI =====")
for kata in ["tidak", "bukan", "kurang", "gak"]:
    jumlah = df["text"].str.contains(
        rf"\b{kata}\b", case=False, na=False
    ).sum()
    print(f"{kata}: {jumlah} posting")

print("\n===== 5. JUMLAH AKHIR =====")
print("Total baris:", len(df))
print("Kolom:", df.columns.tolist())
```
